## Ejercicio 1 — Clasificar el stock de un producto

Se declara la cantidad en inventario de un producto y se usa una cadena `if / elif / else` para clasificar el stock en "Agotado" (cantidad igual a 0), "Stock bajo" (menor a 5) o "Stock suficiente" (en cualquier otro caso). El resultado se guarda en la variable `nivel` y se muestra con `print()`.

```python
cantidad_en_stock = 3

if cantidad_en_stock == 0:
    nivel = "Agotado"
elif cantidad_en_stock < 5:
    nivel = "Stock bajo"
else:
    nivel = "Stock suficiente"

print(nivel)
```

## Ejercicio 2 — Alerta de reposición inmediata

Retomando `cantidad_en_stock` y `nivel` del Ejercicio 1, se usa `if / else` para decidir si se necesita reponer el producto: si el nivel es "Agotado" o la cantidad es menor a 5, se imprime "Reponer pronto"; en cualquier otro caso, se imprime "Sin acción requerida".

```python
if nivel == "Agotado" or cantidad_en_stock < 5:
    print("Reponer pronto")
else:
    print("Sin acción requerida")
```

## Ejercicio 3 — Clasificar el clima del lunes

Se declara la temperatura del lunes y se usa una cadena `if / elif / else` para clasificar el clima en "Frío" (menor a 15), "Templado" (entre 15 y 25 inclusive) o "Caluroso" (mayor a 25). El resultado se guarda en `clima` y se muestra con `print()`.

```python
temp_lunes = 18.5

if temp_lunes < 15:
    clima = "Frío"
elif temp_lunes <= 25:
    clima = "Templado"
else:
    clima = "Caluroso"

print(clima)
```

## Ejercicio 5 — Aprobación con mensaje

Se declaran tres notas parciales, se calcula su promedio, y se usa una cadena `if / elif / else` para clasificar el resultado en "Reprobado" (promedio menor a 3.0), "Aprobado" (entre 3.0 y 4.4 inclusive) o "Aprobado con excelencia" (4.5 o más). El resultado se guarda en `resultado` y se muestra con `print()`.

```python
nota_1 = 4.2
nota_2 = 3.5
nota_3 = 4.8

promedio_notas = (nota_1 + nota_2 + nota_3) / 3

if promedio_notas < 3.0:
    resultado = "Reprobado"
elif promedio_notas <= 4.4:
    resultado = "Aprobado"
else:
    resultado = "Aprobado con excelencia"

print(resultado)
```

## Ejercicio 6 — Recorrer el inventario completo

Se declara una lista con las cantidades en stock de varios productos. Con un `for` que usa `range(len(...))`, se recorre la lista por índice, se clasifica cada cantidad con la misma lógica del Ejercicio 1, y se muestra con `print()` un mensaje indicando el índice del producto y el nivel obtenido.

```python
cantidades_en_stock = [0, 3, 12, 0, 7, 2]

for indice in range(len(cantidades_en_stock)):
    cantidad = cantidades_en_stock[indice]

    if cantidad == 0:
        nivel = "Agotado"
    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    print(f"Producto {indice}: {nivel}")
```

## Ejercicio 7 — Promedio semanal de temperaturas

Se declara una lista con las temperaturas de la semana, junto con un acumulador `suma_temperaturas` y un contador `dias_calurosos`, ambos en 0. Con un `for` que usa `range(len(...))`, se suma cada temperatura al acumulador y se incrementa el contador cuando la temperatura del día supera 20 grados. Después del `for`, se calcula el promedio semanal dividiendo el acumulador entre la cantidad de días, y se muestran ambos resultados con `print()`.

```python
temperaturas_semana = [18.5, 21.0, 19.8, 24.3, 17.2]
suma_temperaturas = 0
dias_calurosos = 0

for indice in range(len(temperaturas_semana)):
    suma_temperaturas = suma_temperaturas + temperaturas_semana[indice]

    if temperaturas_semana[indice] > 20:
        dias_calurosos = dias_calurosos + 1

promedio_semana = suma_temperaturas / len(temperaturas_semana)

print(promedio_semana)
print(dias_calurosos)
```

## Ejercicio 8 — Comprar tornillos mientras alcance el presupuesto

Se declara el presupuesto disponible en la ferretería y el precio de un tornillo, junto con un contador `tornillos_comprados` en 0. Con un `while`, se repite mientras el presupuesto sea mayor o igual al precio del tornillo: en cada vuelta se descuenta el precio del presupuesto y se incrementa el contador. Después del `while`, se muestra con `print()` cuántos tornillos se compraron y cuánto presupuesto quedó sin usar.

```python
presupuesto_ferreteria = 3000
precio_tornillo = 250
tornillos_comprados = 0

while presupuesto_ferreteria >= precio_tornillo:
    presupuesto_ferreteria = presupuesto_ferreteria - precio_tornillo
    tornillos_comprados = tornillos_comprados + 1

print(tornillos_comprados)
print(presupuesto_ferreteria)
```

## Ejercicio 9 — Buscar la primera herramienta cara

Se declara una lista con los precios de varias herramientas. Con un `for` que recorre la lista directamente por sus valores, se revisa cada precio: si supera 20000, se muestra con `print()` un mensaje indicando que se encontró la primera herramienta cara junto con su precio, y se corta el bucle de inmediato con `break`. Si el precio no supera 20000, se muestra un mensaje indicando que se sigue revisando, con el precio actual.

```python
precios_herramientas = [12000.0, 8500.0, 25000.0, 9800.0]

for precio in precios_herramientas:
    if precio > 20000:
        print(f"Primera herramienta cara encontrada, precio: {precio}")
        break
    else:
        print(f"Sigue revisando, precio actual: {precio}")
```

## Ejercicio 10 — Revisión completa del inventario de la ferretería

Se declaran dos listas paralelas con los nombres y las cantidades de varias herramientas. Con un `for` que usa `range(len(...))`, se recorren ambas listas por índice y se clasifica la cantidad de cada posición en "Agotado", "Stock bajo" o "Stock suficiente". Si el nivel es "Stock suficiente", se usa `continue` para saltar esa herramienta sin imprimir nada. Si el nivel es "Agotado", se muestra una alerta urgente con el nombre de la herramienta y se usa `break` para detener la revisión de inmediato. Si el nivel es "Stock bajo", se muestra el nombre y la cantidad, y el bucle continúa con normalidad.

```python
nombres_herramientas = ["Martillo", "Taladro", "Sierra", "Destornillador", "Nivel", "Llave inglesa"]
cantidades_herramientas = [5, 0, 2, 8, 0, 1]

for indice in range(len(nombres_herramientas)):
    cantidad = cantidades_herramientas[indice]
    nombre = nombres_herramientas[indice]

    if cantidad == 0:
        nivel = "Agotado"
    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    if nivel == "Stock suficiente":
        continue
    elif nivel == "Agotado":
        print(f"Alerta: reposición urgente de {nombre}")
        break
    else:
        print(f"{nombre}: {cantidad}")
```
