## Ejercicio 1

Se hacen las variables básicas del inventar
El nombre del producto, los precios, los descuentos , y las cantidades a comprar.


```python
nombre_producto = "Cuaderno"
precio_cuaderno = 3500
precio_lapiz = 800
precio_cafe = 6200.50
hay_descuento = False
cantidad_cuadernos = 3
cantidad_lapices = 5

print(type(nombre_producto))
print(type(precio_cuaderno))
print(type(precio_lapiz))
print(type(precio_cafe))
print(type(hay_descuento))
print(type(cantidad_cuadernos))
print(type(cantidad_lapices))
```

## Ejercicio 2

Se calcula el costo total de los productos comprados, el total general de la compra y el total con un descuento del 10% aplicado sobre ese total general.

```python
total_cuadernos = precio_cuaderno * cantidad_cuadernos
print(total_cuadernos)

total_lapices = precio_lapiz * cantidad_lapices
print(total_lapices)

total_general = total_cuadernos + total_lapices + precio_cafe
print(total_general)

descuento = total_general * 0.10
print(descuento)

total_con_descuento = total_general - descuento
print(total_con_descuento)
```

## Ejercicio 10

Se tienen dos repuestos de ferretería cuyos precios quedaron intercambiados por error. Se muestran los valores antes del intercambio, se corrige el error usando una variable temporal para intercambiar valores.

```python
precio_tornillo = 250
precio_tuerca = 180

print(precio_tornillo)
print(precio_tuerca)

temporal = precio_tornillo
precio_tornillo = precio_tuerca
precio_tuerca = temporal

print(precio_tornillo)
print(precio_tuerca)
```
