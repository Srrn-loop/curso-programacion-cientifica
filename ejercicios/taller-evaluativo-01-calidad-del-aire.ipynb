## Ejercicio 2 — Conversión de unidades y error frente al equipo patrón

Se declara la lectura del sensor de campo (en mg/m³) y el valor de referencia del equipo patrón (en µg/m³). La lectura se convierte a µg/m³ multiplicándola por `factor_conversion` del Ejercicio 1. Con la lectura ya convertida se calcula el error absoluto frente al patrón, el error relativo porcentual y el error cuadrático, usando multiplicación, resta, división y potencia.

```
lectura_sensor_mg = 0.0418
valor_patron_ug = 40.0

lectura_sensor_ug = lectura_sensor_mg * factor_conversion
error_absoluto = lectura_sensor_ug - valor_patron_ug
error_relativo_porcentual = (error_absoluto / valor_patron_ug) * 100
error_cuadratico = error_absoluto ** 2

print(lectura_sensor_ug)
print(error_absoluto)
print(error_relativo_porcentual)
print(error_cuadratico)
```

## Ejercicio 3 — Cobertura temporal del lote descargado

## Ejercicio 6 — Limpiar la serie horaria

Se declara una serie de doce lecturas consecutivas con dos fallas de sensor marcadas con el centinela -999.0. Recorriendo la lista directamente con un `for`, cada lectura igual al centinela se descarta con `continue` (incrementando un contador de descartes), y las demás se agregan a una lista de lecturas válidas y se suman en un acumulador. Después del `for`, se calcula el promedio dividiendo la suma entre la cantidad de lecturas válidas.

```
lecturas_pm25 = [8.4, 12.7, -999.0, 15.2, 41.8, 9.6, -999.0, 58.3, 22.0, 13.9, 7.5, 36.1]

lecturas_validas = []
suma_pm25 = 0.0
lecturas_descartadas = 0

for valor in lecturas_pm25:
    if valor == -999.0:
        lecturas_descartadas = lecturas_descartadas + 1
        continue

    lecturas_validas.append(valor)
    suma_pm25 = suma_pm25 + valor

promedio_pm25 = suma_pm25 / len(lecturas_validas)

print(len(lecturas_validas))
print(lecturas_descartadas)
print(promedio_pm25)
```

## Ejercicio 7 — Estadística descriptiva de la serie limpia

Con `lecturas_validas` y `promedio_pm25` del Ejercicio 6, se calculan los extremos de la serie recorriéndola una vez con `for` y actualizando el máximo y el mínimo con `if` cuando corresponde, sin usar `max()` ni `min()`. Con el rango ya calculado, se recorre la serie una segunda vez para acumular la suma de las diferencias al cuadrado respecto al promedio, y con ella se calcula la desviación estándar usando el operador de potencia elevado a 0.5, sin importar ninguna librería.

```
maximo_pm25 = lecturas_validas[0]
minimo_pm25 = lecturas_validas[0]

for valor in lecturas_validas:
    if valor > maximo_pm25:
        maximo_pm25 = valor
    if valor < minimo_pm25:
        minimo_pm25 = valor

rango_pm25 = maximo_pm25 - minimo_pm25

suma_desviaciones = 0.0

for valor in lecturas_validas:
    suma_desviaciones = suma_desviaciones + (valor - promedio_pm25) ** 2

desviacion_estandar = (suma_desviaciones / len(lecturas_validas)) ** 0.5

print(maximo_pm25)
print(minimo_pm25)
print(rango_pm25)
print(desviacion_estandar)
```

## Ejercicio 8 — Decaimiento tras cerrar la fuente de emisión

Se declara la concentración pico registrada, el factor de decaimiento horario y el umbral de la categoría "Buena". Con un `while`, se repite mientras la concentración siga por encima del umbral: en cada vuelta se multiplica la concentración por el factor de decaimiento y se incrementa el contador de horas transcurridas. Después del `while`, se muestra cuántas horas pasaron y con qué concentración final se cerró el episodio.

```
concentracion = 58.3
factor_decaimiento = 0.85
umbral_buena = 12.0
horas_transcurridas = 0

while concentracion > umbral_buena:
    concentracion = concentracion * factor_decaimiento
    horas_transcurridas = horas_transcurridas + 1

print(horas_transcurridas)
print(concentracion)
```

## Ejercicio 9 — Registros de varias estaciones

Se declara una lista de diccionarios, uno por estación, cada uno con su código, sector, coordenadas (como tupla) y su lista de lecturas; solo una estación tiene el dato de humedad relativa. Recorriendo la lista con un `for`, se desempaqueta la tupla de coordenadas en latitud y longitud, se calcula el promedio de las lecturas de esa estación con un `for` interno, se obtiene la humedad con `get` y un valor por defecto (para no fallar en las estaciones sin ese sensor), y se clasifica el promedio con la misma cadena `if / elif / else` del Ejercicio 5. Cada dato se accede por su clave, nunca por posición.

```
registros_estaciones = [
    {"codigo": "EST-01", "sector": "Norte",  "coordenadas": (6.28, -75.57), "lecturas": [8.4, 12.7, 15.2]},
    {"codigo": "EST-02", "sector": "Centro", "coordenadas": (6.24, -75.58), "lecturas": [41.8, 58.3, 36.1]},
    {"codigo": "EST-03", "sector": "Sur",    "coordenadas": (6.19, -75.59), "lecturas": [9.6, 13.9, 7.5], "humedad_relativa": 68.0},
]

for estacion in registros_estaciones:
    latitud_estacion, longitud_estacion = estacion["coordenadas"]

    suma_lecturas_estacion = 0.0
    for valor in estacion["lecturas"]:
        suma_lecturas_estacion = suma_lecturas_estacion + valor
    promedio_estacion = suma_lecturas_estacion / len(estacion["lecturas"])

    humedad_estacion = estacion.get("humedad_relativa", "no disponible")

    if promedio_estacion <= 12.0:
        categoria_estacion = "Buena"
    elif promedio_estacion <= 35.4:
        categoria_estacion = "Moderada"
    elif promedio_estacion <= 55.4:
        categoria_estacion = "Dañina para grupos sensibles"
    else:
        categoria_estacion = "Dañina para la salud"

    print(f"{estacion['codigo']} | {estacion['sector']} | lat {latitud_estacion} | lon {longitud_estacion} | promedio {promedio_estacion:.2f} | {categoria_estacion} | humedad {humedad_estacion}")
```
